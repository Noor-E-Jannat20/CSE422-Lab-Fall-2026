```python
"""
A* Search - CSE422 CP1 Solution
===============================

Reads a directed, weighted graph together with a heuristic value for every
vertex, then uses the A* algorithm to find the least-cost path from a start
vertex to a goal vertex.

A* expands vertices in order of  f(n) = g(n) + h(n),  where
    g(n) = the actual cost of the best path found so far from start to n
    h(n) = the heuristic estimate of the remaining cost from n to the goal

The first time the goal is removed from the frontier, the path found is
guaranteed to be optimal -- as long as the heuristic is admissible
(it never overestimates the true remaining cost to the goal).
"""

import heapq


# ---------------------------------------------------------------------------
# 1. Input parsing
# ---------------------------------------------------------------------------

def parse_input(text):
    """
    Parse the problem input, which is given as a single multi-line string.

    Expected format:
        line 1        : n m         (number of vertices, number of edges)
        line 2        : start goal  (start vertex, goal vertex)
        next n lines  : vertex h    (a vertex and its heuristic value)
        next m lines  : u v w       (a directed edge u -> v with weight w)

    Returns a tuple: (start, goal, heuristics, edges)
        start, goal : str
        heuristics  : dict mapping  vertex -> int heuristic value
        edges       : list of (u, v, w) tuples, where w is an int weight
    """
    # Keep only the non-empty lines, so stray blank lines in the pasted
    # input do not break the parsing.
    lines = [line.strip() for line in text.strip().splitlines() if line.strip()]

    # Line 1: number of vertices (n) and number of edges (m).
    n, m = map(int, lines[0].split())

    # Line 2: the start vertex and the goal vertex.
    start, goal = lines[1].split()

    # The next n lines hold the heuristic value of each vertex.
    heuristics = {}
    for line in lines[2:2 + n]:
        vertex, h_value = line.split()
        heuristics[vertex] = int(h_value)

    # The following m lines hold the directed, weighted edges.
    edges = []
    for line in lines[2 + n:2 + n + m]:
        u, v, w = line.split()
        edges.append((u, v, int(w)))

    return start, goal, heuristics, edges

```

```python

# ---------------------------------------------------------------------------
# 2. Graph construction
# ---------------------------------------------------------------------------

def build_graph(vertices, edges):
    """
    Build an adjacency list for a directed, weighted graph.

    vertices : an iterable of all vertex names. Passing every vertex (not just
               the ones that appear as an edge source) guarantees that sink
               vertices with no outgoing edges still exist in the graph.
    edges    : list of (u, v, w) tuples meaning a directed edge u -> v of weight w.

    Returns a dict mapping  each vertex -> list of (neighbour, weight) pairs.
    """
    # Give every vertex an empty neighbour list to begin with.
    graph = {vertex: [] for vertex in vertices}

    # Add each directed edge to its source vertex's neighbour list.
    for u, v, w in edges:
        graph[u].append((v, w))

    return graph


```

```python
# ---------------------------------------------------------------------------
# 3. A* search
# ---------------------------------------------------------------------------

def astar_search(graph, heuristics, start, goal):
    """
    Find the least-cost path from start to goal using A* search.

    graph      : adjacency list  (vertex -> list of (neighbour, weight)).
    heuristics : dict mapping  vertex -> heuristic estimate to the goal.
    start, goal: the start and goal vertices.

    Returns (path, cost):
        path : list of vertices from start to goal, or None if unreachable.
        cost : total cost of that path, or None if unreachable.
    """
    # write your code here
    pList=[]
    parent=[]
    cost=0

    heapq.heapify(pList)

    heapq.heappush(pList,graph[start]) #pushing the start goal

    while pList:
      u,vw_list=heapq.heappop(pList)
      parent.append(u)

      if u==goal:
        path=parent.reverse()

      for i in vw_list:
        h= heuristics[u]
        g= i[1]
        f= h+g

        heapq.heappush(pList,graph[i[0]])



    return path, cost
```

```python
# ---------------------------------------------------------------------------
# 4. Driver
# ---------------------------------------------------------------------------

# The sample input, exactly as it appears in the problem statement.
SAMPLE_INPUT = """
7 9
A G
A 5
B 6
C 4
D 3
E 3
F 1
G 0
A B 1
A C 4
B C 2
B D 3
C E 5
D F 2
D G 4
E G 3
F G 1
"""


def main():
    """Parse the input, run A*, and print the path and its total cost."""
    start, goal, heuristics, edges = parse_input(SAMPLE_INPUT)
    graph = build_graph(heuristics.keys(), edges)

    path, cost = astar_search(graph, heuristics, start, goal)

    if path is None:
        print("No path")
    else:
        print(" ".join(path))   # the path, e.g. "A B D F G"
        print(cost)             # the total cost of that path


if __name__ == "__main__":
    main()
```
